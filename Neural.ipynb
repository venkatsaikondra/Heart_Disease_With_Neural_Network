```python
def cost(y_hat, y):
    """
    Binary cross-entropy cost
    y_hat: predicted probabilities (1, m)
    y: true labels (1, m)
    """
    m = y.shape[1]  # number of examples
    epsilon = 1e-8  # prevent log(0)

    # make sure y and y_hat are NumPy arrays
    y_hat = np.clip(y_hat, epsilon, 1 - epsilon)

    # compute cost
    losses = -(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))
    cost_value = np.sum(losses) / m

    return np.squeeze(cost_value)  # ensures it's a scalar
```

```python
import numpy as np

# ---------- ACTIVATION FUNCTIONS ----------
def relu(Z):
    return np.maximum(0, Z)

def relu_derivative(Z):
    return (Z > 0).astype(float)

def sigmoid(Z):
    return 1 / (1 + np.exp(-Z))


# ---------- FEED FORWARD ----------
def feed_forward(A0):
    Z1 = W1 @ A0 + b1
    A1 = relu(Z1)

    Z2 = W2 @ A1 + b2
    A2 = relu(Z2)

    Z3 = W3 @ A2 + b3
    A3 = sigmoid(Z3)

    cache = {"A0": A0, "A1": A1, "A2": A2, "A3": A3, "Z1": Z1, "Z2": Z2, "Z3": Z3}
    return A3, cache


# ---------- COST FUNCTION ----------
def cost(y_hat, Y):
    m = Y.shape[1]
    epsilon = 1e-8
    loss = -(1/m) * np.sum(Y * np.log(y_hat + epsilon) + (1 - Y) * np.log(1 - y_hat + epsilon))
    return np.squeeze(loss)


# ---------- BACKPROP LAYER 3 ----------
def backprop_layer_3(y_hat, Y, m, A2, W3):
    dC_dZ3 = (1/m) * (y_hat - Y)
    dC_dW3 = dC_dZ3 @ A2.T
    dC_db3 = np.sum(dC_dZ3, axis=1, keepdims=True)
    dC_dA2 = W3.T @ dC_dZ3
    return dC_dW3, dC_db3, dC_dA2


# ---------- BACKPROP LAYER 2 ----------
def backprop_layer_2(propagator_dC_dA2, A1, A2, W2, Z2):
    dA2_dZ2 = relu_derivative(Z2)
    dC_dZ2 = propagator_dC_dA2 * dA2_dZ2
    dC_dW2 = dC_dZ2 @ A1.T
    dC_db2 = np.sum(dC_dZ2, axis=1, keepdims=True)
    dC_dA1 = W2.T @ dC_dZ2
    return dC_dW2, dC_db2, dC_dA1


# ---------- BACKPROP LAYER 1 ----------
def backprop_layer_1(propagator_dC_dA1, A1, A0, W1, Z1):
    dA1_dZ1 = relu_derivative(Z1)
    dC_dZ1 = propagator_dC_dA1 * dA1_dZ1
    dC_dW1 = dC_dZ1 @ A0.T
    dC_db1 = np.sum(dC_dZ1, axis=1, keepdims=True)
    return dC_dW1, dC_db1


# ---------- TRAIN FUNCTION ----------
def train(A0, Y, epochs=1000, alpha=0.1):
    global W1, b1, W2, b2, W3, b3
    m = Y.shape[1]
    costs = []

    for e in range(epochs):
        # 1. FEED FORWARD
        y_hat, cache = feed_forward(A0)

        # 2. COST
        error = cost(y_hat, Y)
        costs.append(error)

        # 3. BACKPROP
        dC_dW3, dC_db3, dC_dA2 = backprop_layer_3(y_hat, Y, m, cache["A2"], W3)
        dC_dW2, dC_db2, dC_dA1 = backprop_layer_2(dC_dA2, cache["A1"], cache["A2"], W2, cache["Z2"])
        dC_dW1, dC_db1 = backprop_layer_1(dC_dA1, cache["A1"], cache["A0"], W1, cache["Z1"])

        # 4. UPDATE WEIGHTS
        W3 -= alpha * dC_dW3
        b3 -= alpha * dC_db3
        W2 -= alpha * dC_dW2
        b2 -= alpha * dC_db2
        W1 -= alpha * dC_dW1
        b1 -= alpha * dC_db1

        # 5. PRINT EVERY 100 EPOCHS
        if e % 100 == 0:
            print(f"Epoch {e}: Cost = {error:.4f}")

    return costs


# ---------- ACCURACY ----------
def accuracy(X, Y):
    y_hat, _ = feed_forward(X)
    predictions = (y_hat > 0.5).astype(int)
    acc = np.mean(predictions == Y)
    return acc
```

```python
def cost(y_hat, y):
    """
    Binary cross-entropy cost
    y_hat: predicted probabilities (1, m)
    y: true labels (1, m)
    """
    m = y.shape[1]  # number of examples
    epsilon = 1e-8  # prevent log(0)

    # make sure y and y_hat are NumPy arrays
    y_hat = np.clip(y_hat, epsilon, 1 - epsilon)

    # compute cost
    losses = -(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))
    cost_value = np.sum(losses) / m

    return np.squeeze(cost_value)  # ensures it's a scalar
```

```python
def backprop(Y, cache):
    m = Y.shape[1]
    A0, A1, A2, A3, A4 = cache["A0"], cache["A1"], cache["A2"], cache["A3"], cache["A4"]
    Z1, Z2, Z3 = cache["Z1"], cache["Z2"], cache["Z3"]

    dZ4 = A4 - Y
    dW4 = (1/m) * (dZ4 @ A3.T)
    db4 = (1/m) * np.sum(dZ4, axis=1, keepdims=True)

    dA3 = W4.T @ dZ4
    dZ3 = dA3 * leaky_relu_derivative(Z3)
    dW3 = (1/m) * (dZ3 @ A2.T)
    db3 = (1/m) * np.sum(dZ3, axis=1, keepdims=True)

    dA2 = W3.T @ dZ3
    dZ2 = dA2 * leaky_relu_derivative(Z2)
    dW2 = (1/m) * (dZ2 @ A1.T)
    db2 = (1/m) * np.sum(dZ2, axis=1, keepdims=True)

    dA1 = W2.T @ dZ2
    dZ1 = dA1 * leaky_relu_derivative(Z1)
    dW1 = (1/m) * (dZ1 @ A0.T)
    db1 = (1/m) * np.sum(dZ1, axis=1, keepdims=True)

    grads = {"dW4": dW4, "db4": db4,
             "dW3": dW3, "db3": db3,
             "dW2": dW2, "db2": db2,
             "dW1": dW1, "db1": db1}

    return grads
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# --------------------------------
# 1. Load and preprocess the data
# --------------------------------
# Assume df is already loaded, e.g.:
# df = pd.read_csv("heart.csv")

# Encode categorical columns
df_encoded = pd.get_dummies(df, columns=['cp', 'restecg', 'slope', 'thal'], drop_first=False)

# Separate input and output
X = df_encoded.drop('target', axis=1).values  # numpy array
y = df_encoded['target'].values.reshape(1, -1)  # shape (1, m)

# Normalize numeric features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, y.T, test_size=0.3, random_state=42, shuffle=True)

# Transpose so each column is one example
X_train, X_test = X_train.T, X_test.T
Y_train, Y_test = Y_train.T, Y_test.T

# Dimensions
n_x = X_train.shape[0]
n_h1, n_h2, n_h3 = 64, 32, 16
n_y = 1
m = X_train.shape[1]

# --------------------------------
# 2. Activation & Utility Functions
# --------------------------------
def leaky_relu(Z, alpha=0.01):
    return np.where(Z > 0, Z, alpha * Z)

def leaky_relu_derivative(Z, alpha=0.01):
    dZ = np.ones_like(Z)
    dZ[Z < 0] = alpha
    return dZ

def sigmoid(Z):
    return 1 / (1 + np.exp(-Z))

def sigmoid_derivative(A):
    return A * (1 - A)

def cost(y_hat, Y):
    m = Y.shape[1]
    epsilon = 1e-8
    return -(1/m) * np.sum(Y * np.log(y_hat + epsilon) + (1 - Y) * np.log(1 - y_hat + epsilon))

def accuracy(y_hat, Y):
    preds = (y_hat > 0.5).astype(int)
    return np.mean(preds == Y) * 100

# --------------------------------
# 3. Initialize weights (He init)
# --------------------------------
np.random.seed(42)
W1 = np.random.randn(n_h1, n_x) * np.sqrt(2/n_x)
b1 = np.zeros((n_h1, 1))

W2 = np.random.randn(n_h2, n_h1) * np.sqrt(2/n_h1)
b2 = np.zeros((n_h2, 1))

W3 = np.random.randn(n_h3, n_h2) * np.sqrt(2/n_h2)
b3 = np.zeros((n_h3, 1))

W4 = np.random.randn(n_y, n_h3) * np.sqrt(2/n_h3)
b4 = np.zeros((n_y, 1))

# --------------------------------
# 4. Feed Forward
# --------------------------------
def feed_forward(X):
    cache = {}

    Z1 = W1 @ X + b1
    A1 = leaky_relu(Z1)

    Z2 = W2 @ A1 + b2
    A2 = leaky_relu(Z2)

    Z3 = W3 @ A2 + b3
    A3 = leaky_relu(Z3)

    Z4 = W4 @ A3 + b4
    A4 = sigmoid(Z4)

    cache = {"A0": X, "Z1": Z1, "A1": A1,
              "Z2": Z2, "A2": A2, "Z3": Z3, "A3": A3,
              "Z4": Z4, "A4": A4}

    return A4, cache

# --------------------------------
# 5. Backpropagation
# --------------------------------
def backprop(Y, cache):
    m = Y.shape[1]
    A0, A1, A2, A3, A4 = cache["A0"], cache["A1"], cache["A2"], cache["A3"], cache["A4"]
    Z1, Z2, Z3 = cache["Z1"], cache["Z2"], cache["Z3"]

    dZ4 = A4 - Y
    dW4 = (1/m) * (dZ4 @ A3.T)
    db4 = (1/m) * np.sum(dZ4, axis=1, keepdims=True)

    dA3 = W4.T @ dZ4
    dZ3 = dA3 * leaky_relu_derivative(Z3)
    dW3 = (1/m) * (dZ3 @ A2.T)
    db3 = (1/m) * np.sum(dZ3, axis=1, keepdims=True)

    dA2 = W3.T @ dZ3
    dZ2 = dA2 * leaky_relu_derivative(Z2)
    dW2 = (1/m) * (dZ2 @ A1.T)
    db2 = (1/m) * np.sum(dZ2, axis=1, keepdims=True)

    dA1 = W2.T @ dZ2
    dZ1 = dA1 * leaky_relu_derivative(Z1)
    dW1 = (1/m) * (dZ1 @ A0.T)
    db1 = (1/m) * np.sum(dZ1, axis=1, keepdims=True)

    grads = {"dW4": dW4, "db4": db4,
             "dW3": dW3, "db3": db3,
             "dW2": dW2, "db2": db2,
             "dW1": dW1, "db1": db1}

    return grads

# --------------------------------
# 6. Training function
# --------------------------------
def train(X, Y, epochs=2000, alpha=0.01):
    global W1, W2, W3, W4, b1, b2, b3, b4
    costs = []

    for e in range(epochs):
        y_hat, cache = feed_forward(X)
        loss = cost(y_hat, Y)
        grads = backprop(Y, cache)

        # Gradient descent update
        W1 -= alpha * grads["dW1"]
        b1 -= alpha * grads["db1"]
        W2 -= alpha * grads["dW2"]
        b2 -= alpha * grads["db2"]
        W3 -= alpha * grads["dW3"]
        b3 -= alpha * grads["db3"]
        W4 -= alpha * grads["dW4"]
        b4 -= alpha * grads["db4"]

        if e % 100 == 0:
            acc = accuracy(y_hat, Y)
            print(f"Epoch {e}: Cost = {loss:.4f}, Train Accuracy = {acc:.2f}%")
        costs.append(loss)

    return costs

# --------------------------------
# 7. Train and Evaluate
# --------------------------------
costs = train(X_train, Y_train, epochs=2000, alpha=0.01)

# Predictions
y_hat_train, _ = feed_forward(X_train)
y_hat_test, _ = feed_forward(X_test)

train_acc = accuracy(y_hat_train, Y_train)
test_acc = accuracy(y_hat_test, Y_test)

print(f"\n✅ Final Training Accuracy: {train_acc:.2f}%")
print(f"✅ Final Test Accuracy: {test_acc:.2f}%")
```
